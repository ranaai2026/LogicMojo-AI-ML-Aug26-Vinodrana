### Problem-1

In a network, data packets are sent through a shared channel. Suppose there are 10 devices attempting to send data. In any given time slot, each device has a 20% chance of attempting to send a packet, independently of other devices. A collision occurs if two or more devices attempt to send a packet in the same time slot. What is the probability that in a given time slot, there will be no collisions?

Problem 1 — Probability of No Collision

There are 10 devices, and each device independently attempts to send with probability:

$$ p=0.20 $$

Therefore, the probability that a device does not attempt is:

$$ q=1-0.20=0.80 $$

A collision occurs when 2 or more devices attempt simultaneously.

So, no collision means either:

No device attempts, or
Exactly one device attempts.
Case 1: No device attempts
$$ P(0)=\binom{10}{0}(0.2)^0(0.8)^{10} $$ $$ P(0)=0.8^{10}=0.1073741824 $$
Case 2: Exactly one device attempts
$$ P(1)=\binom{10}{1}(0.2)^1(0.8)^9 $$ $$ P(1)=10(0.2)(0.8)^9 $$ $$ P(1)=0.268435456 $$

Therefore:

$$ P(\text{no collision})=P(0)+P(1) $$ $$ =0.1073741824+0.268435456 $$ $$ \boxed{P(\text{no collision})=0.3758096384} $$

So the probability is approximately:

37.58%
	​


````python
import math

n = 10
p = 0.20
q = 1 - p

prob_no_collision = (q ** n) + (n * p * q ** (n - 1))

print(prob_no_collision)
print(prob_no_collision * 100)
````

### Problem-7

A batch of 500 microchips is produced. The probability that a single microchip is defective is 0.02.
Let X be the number of defective microchips in a random sample of 20 microchips from this batch. Assume the sample is taken with replacement, or the batch is large enough that we can model this as independent trials.

(a) What is the probability that exactly 3 microchips in the sample are defective?
(b) What is the probability that at least 2 microchips in the sample are defective?

Binomial Distribution

We have:

Number of trials:
$$ n=20 $$
Probability of a defective chip:
$$ p=0.02 $$
Probability of a non-defective chip:
$$ q=1-p=0.98 $$

Since each trial has two possible outcomes—defective or not defective—we use the binomial distribution.

The formula is:

$$ P(X=k)=\binom nk p^k(1-p)^{n-k} $$
(a) Exactly 3 defective chips

Here:

$$ k=3 $$

Therefore:

$$ P(X=3)=\binom{20}{3}(0.02)^3(0.98)^{17} $$

Calculate:

$$ \binom{20}{3} =\frac{20!}{3!17!} =1140 $$

Thus:

$$ P(X=3)=1140(0.02)^3(0.98)^{17} $$

Using Python or a calculator:

$$ \boxed{P(X=3)\approx0.00707} $$

or approximately:

0.707%
	​


````python
import math

n = 20
p = 0.02
k = 3

prob_3 = math.comb(n, k) * (p ** k) * ((1-p) ** (n-k))

print(prob_3)
print(prob_3 * 100)
````

(b) At least 2 defective chips

"At least 2" means:

$$ X\ge2 $$

We could calculate:

$$ P(X=2)+P(X=3)+\cdots+P(X=20) $$

But that's tedious.

Instead, use the complement:

$$ P(X\ge2)=1-P(X<2) $$

Since \(X<2\) means \(X=0\) or \(X=1\):

$$ P(X\ge2)=1-[P(X=0)+P(X=1)] $$
Calculate \(P(X=0)\)
$$ P(X=0)=\binom{20}{0}(0.02)^0(0.98)^{20} $$ $$ \approx0.66761 $$
Calculate \(P(X=1)\)
$$ P(X=1)=\binom{20}{1}(0.02)^1(0.98)^{19} $$ $$ \approx0.27248 $$

Therefore:

$$ P(X\ge2)=1-(0.66761+0.27248) $$ $$ \boxed{P(X\ge2)\approx0.05991} $$

or approximately:

5.99%
	​


````python
import math

n = 20
p = 0.02

p0 = math.comb(n, 0) * (p ** 0) * ((1-p) ** n)
p1 = math.comb(n, 1) * (p ** 1) * ((1-p) ** (n-1))

prob_at_least_2 = 1 - (p0 + p1)

print(prob_at_least_2)
print(prob_at_least_2 * 100)
````

### Problem-8

The time (in hours) until a critical server component fails follows an exponential distribution with a mean of 500 hours.

(a) What is the probability that the component will last between 400 and 600 hours?

(b) The component has already been running for 300 hours without failure. What is the probability that it will last for at least another 250 hours?

Exponential Distribution

The mean lifetime is:

$$ 500\text{ hours} $$

For an exponential distribution:

$$ \lambda=\frac{1}{\text{mean}} $$

Therefore:

$$ \lambda=\frac{1}{500}=0.002 $$

The survival function is:

$$ P(X>t)=e^{-\lambda t} $$
(a) Probability the component lasts between 400 and 600 hours

We want:

$$ P(400<X<600) $$

This can be calculated as:

$$ P(X<600)-P(X<400) $$

For an exponential distribution:

$$ P(X<t)=1-e^{-\lambda t} $$

Therefore:

$$ P(400<X<600) = e^{-\lambda(400)}-e^{-\lambda(600)} $$

Substitute \(\lambda=0.002\):

$$ =e^{-0.8}-e^{-1.2} $$ $$ \approx0.44933-0.3010 $$ $$ \boxed{P(400<X<600)\approx0.1483} $$

Therefore:

14.83%
	​


````python
import math

lam = 1 / 500

prob_400_600 = math.exp(-lam * 400) - math.exp(-lam * 600)

print(prob_400_600)
print(prob_400_600 * 100)
````

(b) Already running for 300 hours

We want the probability that it lasts at least another 250 hours, given that it has already survived 300 hours:

$$ P(X>550\mid X>300) $$

The exponential distribution has the memoryless property:

$$ P(X>s+t|X>s)=P(X>t) $$

Therefore:

$$ P(X>550|X>300)=P(X>250) $$

Now:

$$ P(X>250)=e^{-\lambda(250)} $$ $$ =e^{-0.002(250)} $$ $$ =e^{-0.5} $$ $$ \boxed{\approx0.60653} $$

Therefore:

$$ \boxed{60.65\%} $$

The important concept is that for an exponential distribution, the previous 300 hours do not change the probability of surviving another 250 hours.

Normal Distribution

$$ Z_2=\frac{0.3-0}{0.5} $$ $$ Z_2=0.6 $$

Therefore:

$$ P(-0.2<X<0.3) = P(-0.4<Z<0.6) $$

Using the standard normal CDF:

$$ \Phi(0.6)\approx0.72575 $$

and

38.12%

### Problem-13

The `rand5()` function which returns a uniform random integer within the range [1, 5], is defined as following:

```python
def rand5():
    import numpy as np
    return np.random.randint(1,6)
```

Write a function `rand7()` that will generate a uniform random integer within the range [1, 7] by only using `rand5()`. 

Note: You can call `rand5()` as many time as you want inside the function `rand7()`, but you can not use `np.random.randint(1, 8)` or any other built-in methods in python. 

We can solve this using rejection sampling. The key is to first generate a uniform number from 1–25 using two calls to rand5(), then reject values 22–25.

Step 1: Generate a number from 1 to 25

Each call to rand5() has 5 equally likely outcomes.

Let:
a=rand5()
b=rand5()
We can combine them as:

$$ ( a-1)\times5+b $$

This produces every integer from 1 to 25, each with probability \(1/25\).

Step 2: Keep only 1–21

Why 21?

Because:

$$ 21=7\times3 $$

So the numbers 1–21 can be divided evenly into 3 groups of 7:

1–7 → 1–7
8–14 → 1–7
15–21 → 1–7

Each number 1–7 will therefore occur exactly 3 times.

If we get 22–25, we simply discard it and try again.

````python
def rand5():
    import numpy as np
    return np.random.randint(1, 6)


def rand7():
    while True:
        a = rand5()
        b = rand5()

        num = (a - 1) * 5 + b

        if num <= 21:
            return ((num - 1) % 7) + 1
````

Why is it uniform?

There are 25 equally likely combinations from two rand5() calls.

We accept only:

1,2,...,21

Each has probability:

1/25

Among those 21 values, each output from 1 to 7 appears exactly 3 times.

For example:

1,8,15\rightarrow 1 
2,9,16\rightarrow 2
...
7,14,21\rightarrow 7

Thus:

P(rand7=k)=3/21=1/7

for every k=1,..,7.

Therefore, rand7() generates a uniform integer from 1 to 7.

### Problem-14

A grayscale image can be represented as a matrix where the dimensions are (height, width). 
Let's consider a simplified scenario with two grayscale images, $A$ and $B$, both represented as 10×10 matrices where each element represents pixel intensity. Pixel intensity is an integer in the range [0-255].

You want to create a blended image $C$ by combining $A$ and $B$ such that $C=w_1 A+w_2 B$. This represents a weighted average blend.
Note that each element in the resultant matrix $C$ should also fall within the range [0, 255]. 

write the code of the function `blend_images(image_A, image_B, weight_A, weight_B)` that takes two matrices and two scalar weights and produces another matrix as an output by performing the above operation.

Yes. Problem 14 is asking you to perform a weighted average of two image matrices.

The blending operation is:

$$ \boxed{\text{Blended Image}=weight_A\times image_A+weight_B\times image_B} $$

Typically, the weights should satisfy:

$$ weight_A+weight_B=1 $$

For example, if weight_A = 0.7 and weight_B = 0.3, the result is 70% image A and 30% image B.

Python solution using NumPy

````python
import numpy as np

def blend_images(image_A, image_B, weight_A, weight_B):
    
    # Check that the weights add up to 1
    if not np.isclose(weight_A + weight_B, 1):
        raise ValueError("weight_A and weight_B must add up to 1")
    
    # Weighted average of the two images
    blended_image = weight_A * image_A + weight_B * image_B
    
    # Keep pixel values within the range [0, 255]
    blended_image = np.clip(blended_image, 0, 255)
    
    # Convert pixel intensities to integers
    blended_image = blended_image.astype(np.uint8)
    
    return blended_image
````

````python
image_A = np.random.randint(0, 256, (10, 10))
image_B = np.random.randint(0, 256, (10, 10))

result = blend_images(image_A, image_B, 0.6, 0.4)

print(result)
print(result.shape)
````

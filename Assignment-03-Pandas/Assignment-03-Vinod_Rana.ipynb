#### 9. How many Counties have 2010 populations greater than 1 million?

```python
# Write your code here
df = pd.read_csv('population_by_county.csv')

# Filter for rows where '2010Census' is strictly greater than 1,000,000
large_counties = df[df['2010Census'] > 1000000]

# Count the number of rows that match the condition
county_count = len(large_counties)

print(f"Number of counties with population > 1 million: {county_count}")
```

#### 10. How many Counties don't have the word "County" in their name?

```python
# Write your code here
df = pd.read_csv('population_by_county.csv')

# Use the ~ operator to invert the string search results
# case=False handles any potential lowercase variations (e.g., 'county' or 'city')
no_county_word = df[~df['County'].str.contains('County', case=False)]

# Calculate total count using len()
count = len(no_county_word)

print(f"Number of counties without the word 'County' in their name: {count}")
```

#### 11. Add a column that calculates the percent change between the 2010 Census and the 2017 Population Estimate

```python
# Write your code here
df = pd.read_csv('population_by_county.csv')

# Add the new column using direct vectorized operations
df['PercentChange'] = ((df['2017PopEstimate'] - df['2010Census']) / df['2010Census']) * 100

# Display the first few rows to verify the new column
print(df[['County', 'State', '2010Census', '2017PopEstimate', 'PercentChange']].head())
```

#### 12. What States have the highest estimated percent change between the 2010 Census and the 2017 Population Estimate?

**NOTE:** This will take several lines of code, as it requires a recalculation of Percent Change.

```python
# Write your code here
df = pd.read_csv('population_by_county.csv')

# Step 1: Aggregate the total population fields by State
state_totals = df.groupby('State')[['2010Census', '2017PopEstimate']].sum()

# Step 2: Recalculate the absolute percent change using the aggregated State totals
state_totals['StatePercentChange'] = (
    (state_totals['2017PopEstimate'] - state_totals['2010Census']) / state_totals['2010Census']
) * 100

# Step 3: Sort the states in descending order and isolate the top 5
top_5_growth_states = state_totals['StatePercentChange'].sort_values(ascending=False).head(5)

print("Top 5 States with the Highest Estimated Percent Change (2010 to 2017):")
print(top_5_growth_states)
```

## Problem: 4

Create a 10 x 10 arrays of zeros and then "frame" it with a border of ones.

```python
# Write your code here
import numpy as np

# Create a 10 x 10 array filled with zeros
array = np.zeros((10, 10), dtype=int)

# Create the border of ones
array[0, :] = 1
array[-1, :] = 1
array[:, 0] = 1
array[:, -1] = 1

print(array)
```

## Problem: 5

Create an 8 x 8 array with a checkerboard pattern of zeros and ones.

```python
# Write your code here
import numpy as np

# Create an 8 x 8 array filled with zeros
array = np.zeros((8, 8), dtype=int)

# Put 1s in alternate positions
array[1::2, ::2] = 1
array[::2, 1::2] = 1

print(array)
```

## Problem: 6

Create an array $x$ with 100 points in the interval $[-1,1]$ (the points must be uniformly spaced). Then create another array $y$. 

Where $y = x^2 sin(x^2) + x$. 

Try to plot $y$ vs $x$ using matplotlib.

```python
# Write your code here
import numpy as np
import matplotlib.pyplot as plt

# Create an array of 100 uniformly spaced points from -1 to 1
x = np.linspace(-1, 1, 100)

# Create y using the given formula
y = x**2 * np.sin(x**2) + x

# Plot y versus x
plt.plot(x, y)

# Add labels and title
plt.xlabel("x")
plt.ylabel("y")
plt.title("Plot of y vs x")

# Display the graph
plt.show()
```
